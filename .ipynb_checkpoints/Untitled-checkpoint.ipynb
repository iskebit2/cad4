```python
def local_collect_fixed_dofs(scene) -> list:
    fixed = []
    for node in scene.nodes.values():
        dofs = getattr(node, "dof_indices", None)
        if not dofs: continue
        r = getattr(node, "restraint", None)
        if r is None: continue
        for i, dof_name in enumerate(("ux", "uy", "uz", "rx", "ry", "rz")):
            if getattr(r, dof_name, False):
                if i < len(dofs) and dofs[i] is not None:
                    fixed.append(int(dofs[i]))
    return fixed
    
def local_stiffness(E, G, A, J, I33, I22, AS2, AS3, L, release_i=None, release_j=None, use_shear_deform=True):
    k = np.zeros((12, 12), dtype=float)
    ri = release_i or {}
    rj = release_j or {}
    def rel(d, code):
        return bool(d.get(code, False))
    if not (rel(ri, "R1") or rel(rj, "R1")):
        EA_L = E * A / L
        k[0, 0] += EA_L; k[6, 6] += EA_L; k[0, 6] -= EA_L; k[6, 0] -= EA_L
    if not (rel(ri, "R4") or rel(rj, "R4")):
        GJ_L = G * J / L
        k[3, 3] += GJ_L; k[9, 9] += GJ_L; k[3, 9] -= GJ_L; k[9, 3] -= GJ_L
    phi3 = 0.0
    if use_shear_deform and G > 0 and AS2 > 0 and I33 > 0:
        phi3 = (12.0 * E * I33) / (G * AS2 * L**2)
    denom3 = 1.0 + phi3
    a3 = 12.0 * E * I33 / (denom3 * L**3)
    b3 = 6.0 * E * I33 / (denom3 * L**2)
    c3 = (4.0 + phi3) * E * I33 / (denom3 * L)
    d3 = (2.0 - phi3) * E * I33 / (denom3 * L)
    k[1, 1] += a3;  k[1, 5] += b3;  k[1, 7] -= a3;  k[1, 11] += b3
    k[5, 1] += b3;  k[5, 5] += c3;  k[5, 7] -= b3;  k[5, 11] += d3
    k[7, 1] -= a3;  k[7, 5] -= b3;  k[7, 7] += a3;  k[7, 11] -= b3
    k[11, 1] += b3; k[11, 5] += d3; k[11, 7] -= b3; k[11, 11] += c3
    phi2 = 0.0
    if use_shear_deform and G > 0 and AS3 > 0 and I22 > 0:
        phi2 = (12.0 * E * I22) / (G * AS3 * L**2)
    denom2 = 1.0 + phi2
    a2 = 12.0 * E * I22 / (denom2 * L**3)
    b2 = 6.0 * E * I22 / (denom2 * L**2)
    c2 = (4.0 + phi2) * E * I22 / (denom2 * L)
    d2 = (2.0 - phi2) * E * I22 / (denom2 * L)
    k[2, 2] += a2;  k[2, 4] -= b2;  k[2, 8] -= a2;  k[2, 10] -= b2
    k[4, 2] -= b2;  k[4, 4] += c2;  k[4, 8] += b2;  k[4, 10] += d2
    k[8, 2] -= a2;  k[8, 4] += b2;  k[8, 8] += a2;  k[8, 10] += b2
    k[10, 2] -= b2; k[10, 4] += d2; k[10, 8] += b2; k[10, 10] += c2
    return 0.5 * (k + k.T)
    
def dynamic_build_stiffness(project, use_shear_deform=True):
    scene = project.scene
    for i, node in enumerate(scene.nodes.values()):
        node.dof_indices = list(range(i * 6, i * 6 + 6))
    num_dofs = len(scene.nodes) * 6
    K = np.zeros((num_dofs, num_dofs), dtype=float)

    for frame in scene.frames.values():
        sec = getattr(frame, "section", None)
        if not sec: continue
        mat = getattr(sec, "material", None)
        if not mat: continue
        params = sec.profile_params or {}
        A = float(params["Area"]) * float(params.get("AMod", 1.0))
        J = float(params["J"]) * float(params.get("JMod", 1.0))
        I33 = float(params["I33"]) * float(params.get("I3Mod", 1.0))
        I22 = float(params["I22"]) * float(params.get("I2Mod", 1.0))
        AS2 = float(params.get("AS2", A))
        AS3 = float(params.get("AS3", A))
        E = float(getattr(mat, "E1", 0.0))
        G = float(getattr(mat, "G12", 0.0))

        p1 = (frame.node_i.x, frame.node_i.y, frame.node_i.z)
        p2 = (frame.node_j.x, frame.node_j.y, frame.node_j.z)

        T, L = transformation_matrix(p1, p2, getattr(frame, "rotation_deg", 0.0) or 0.0)
        k_loc = local_stiffness(E, G, A, J, I33, I22, AS2, AS3, L,
                                getattr(frame, "release_i", None),
                                getattr(frame, "release_j", None),
                                use_shear_deform)
        k_glob = T.T @ k_loc @ T
        k_glob = 0.5 * (k_glob + k_glob.T)
        dofs = list(frame.node_i.dof_indices) + list(frame.node_j.dof_indices)
        K[np.ix_(dofs, dofs)] += k_glob

    K = 0.5 * (K + K.T)
    class TempStiff:
        def __init__(self, K, nd):
            self.K_global = K
            self.num_dofs = nd
    return TempStiff(K, num_dofs)

def transformation_matrix(p1, p2, beta_deg=0.0):
    import math
    p1 = np.asarray(p1, dtype=float)
    p2 = np.asarray(p2, dtype=float)
    v = p2 - p1
    L = float(np.linalg.norm(v))
    if L < 1e-9:
        raise ValueError(f"Sıfır uzunluklu eleman: L={L}")
    e1 = v / L
    dx, dy, dz = e1
    if abs(dz) > 0.9999:
        sign = 1.0 if dz > 0 else -1.0
        e2 = np.array([0.0, sign, 0.0])
        e3 = np.array([-sign, 0.0, 0.0])
    else:
        D = math.hypot(dx, dy)
        e2 = np.array([-dy / D, dx / D, 0.0])
        e3 = np.cross(e1, e2)
    if abs(beta_deg) > 1e-12:
        a = math.radians(beta_deg)
        ca, sa = math.cos(a), math.sin(a)
        e2, e3 = ca * e2 + sa * e3, -sa * e2 + ca * e3
    R = np.vstack([e1, e2, e3])
    T = np.kron(np.eye(4), R)
    return T, L
```

```python
@dataclass
class ModalResult:
    frequencies: np.ndarray            # (n_modes,) Hz
    periods: np.ndarray                # (n_modes,) s
    mode_shapes: np.ndarray            # (n_free_dofs, n_modes)
    modal_dofs: np.ndarray             # (n_free_dofs,) global DOF endeksleri

    modal_mass: np.ndarray             # (n_modes,) genelleştirilmiş kütle
    participation_factor_x: np.ndarray # (n_modes,)
    participation_factor_y: np.ndarray
    participation_factor_z: np.ndarray
    effective_mass_x: np.ndarray       # (n_modes,) kg
    effective_mass_y: np.ndarray
    effective_mass_z: np.ndarray

    total_mass_x: float = 0.0          # kg
    total_mass_y: float = 0.0
    total_mass_z: float = 0.0

    def __len__(self) -> int:
        return len(self.periods)

    @property
    def n_modes(self) -> int:
        return len(self.periods)
        
class StructuralSolver:
    ROT_MASS_SCALE: float = 1e-6
    MIN_MASS: float = 1e-9
    EIGEN_TOL: float = 1e-6

    def __init__(
        self,
        K_global: np.ndarray,
        M_global: np.ndarray,
        fixed_dofs,
        num_dofs: int,
    ):
        self.K = np.asarray(K_global, dtype=float)
        self.M = np.asarray(M_global, dtype=float)
        self.fixed_dofs = {int(d) for d in fixed_dofs if d is not None}
        self.num_dofs = int(num_dofs)

        all_dofs = set(range(self.num_dofs))
        self.free_dofs = np.array(sorted(all_dofs - self.fixed_dofs), dtype=int)

    def _guyan_reduced(self):
        K_red, M_red = self._reduced()
        m_diag = np.diag(M_red)
        master = np.where(m_diag > 1e-9)[0]
        slave = np.where(m_diag <= 1e-9)[0]

        if len(slave) == 0:
            return K_red[np.ix_(master, master)], M_red[np.ix_(master, master)]

        K_red_cond, M_red_cond = perform_guyan_condensation(K_red, M_red, master, slave)
        return K_red_cond, M_red_cond

    def _reduced(self) -> Tuple[np.ndarray, np.ndarray]:
        idx = np.ix_(self.free_dofs, self.free_dofs)
        return self.K[idx], self.M[idx]

    def static_solve(self, F: np.ndarray) -> np.ndarray:
        K_red, _ = self._reduced()
        F_red = np.asarray(F, dtype=float)[self.free_dofs]
        U_red = np.linalg.solve(K_red, F_red)

        U = np.zeros(self.num_dofs)
        U[self.free_dofs] = U_red
        return U

    def modal_solve(
        self,
        n_modes: int = 10,
        debug: bool = True,
        use_condensation: bool = True,
    ) -> ModalResult:
        K_red, M_red = self._reduced()

        dof_types = self.free_dofs % 6
        trans_idx = np.where(dof_types < 3)[0]  # Öteleme (Master / a)
        rot_idx = np.where(dof_types >= 3)[0]    # Rotasyon (Slave / b)

        if debug:
            print(f"\n--- Guyan / Kütlesiz DOF Analizi ---")
            print(f"Toplam serbest DOF: {len(self.free_dofs)}")
            print(f"Öteleme (kütleli) DOF sayısı: {len(trans_idx)}")
            print(f"Rotasyon (kütlesiz) DOF sayısı: {len(rot_idx)}")
            print(f"Kondensasyon: {'AKTİF (Guyan)' if use_condensation else 'KAPALI (Full)'}")

        G: Optional[np.ndarray] = None

        # ==============================================================
        # DAL 1: GUYAN STATİK KONDENSASYONU (Optimize Edilmiş Fonksiyon)
        # ==============================================================
        if use_condensation and len(rot_idx) > 0:
            # Güncel ve kararlı harici kondensasyon alt fonksiyonunu çağırıyoruz
            K_cond, M_cond = perform_guyan_condensation(K_red, M_red, trans_idx, rot_idx)

            # Kinematik ilişki matrisi T_s (G) geri kazanım adımları için hesaplanır
            K_bb = K_red[np.ix_(rot_idx, rot_idx)]
            K_ba = K_red[np.ix_(rot_idx, trans_idx)]
            try:
                from scipy.linalg import cho_factor, cho_solve
                c, low = cho_factor(K_bb, lower=True, check_finite=False)
                G = cho_solve((c, low), -K_ba, check_finite=False)
            except (np.linalg.LinAlgError, ValueError):
                # K_bb tekil veya kötü koşullu ise regularization ekleyerek kararlı çözüm sağlarız
                K_bb_reg = K_bb + 1e-10 * np.eye(K_bb.shape[0])
                try:
                    c, low = cho_factor(K_bb_reg, lower=True, check_finite=False)
                    G = cho_solve((c, low), -K_ba, check_finite=False)
                except Exception:
                    G, *_ = np.linalg.lstsq(K_bb_reg, -K_ba, rcond=None)

            eigenvalues, vectors_a = eigh(K_cond, M_cond)
            eigenvalues = np.maximum(eigenvalues, 0.0)
            order = np.argsort(eigenvalues)
            eigenvalues = eigenvalues[order]
            vectors_a = vectors_a[:, order]

            omega = np.sqrt(eigenvalues)
            frequencies = omega / (2.0 * np.pi)
            valid = frequencies > self.EIGEN_TOL

            eigenvalues = eigenvalues[valid]
            vectors_a = vectors_a[:, valid]
            frequencies = frequencies[valid]
            periods = np.where(frequencies > self.EIGEN_TOL, 1.0 / frequencies, np.inf)

            count = min(n_modes, len(frequencies))
            vectors_a = vectors_a[:, :count]
            frequencies = frequencies[:count]
            periods = periods[:count]

            # Tam DOF uzayına geri dönüştürme (Öteleme + Rotasyon = G @ vectors_a)
            full_vectors = np.zeros((len(self.free_dofs), count))
            full_vectors[trans_idx, :] = vectors_a
            if G is not None:
                full_vectors[rot_idx, :] = G @ vectors_a

            participation, eff_mass, modal_mass = self._participation_condensed(
                vectors_a, M_cond, trans_idx, rot_idx, G, self.free_dofs,
            )
            total_mass_x, total_mass_y, total_mass_z = self._total_masses_condensed(
                M_cond, trans_idx, rot_idx, G, self.free_dofs,
            )

        # ==============================================================
        # DAL 2: FULL DOF ÇÖZÜMÜ
        # ==============================================================
        else:
            M_eff = self._build_effective_mass(M_red)
            eigenvalues, full_vectors = eigh(K_red, M_eff)
            eigenvalues = np.maximum(eigenvalues, 0.0)
            order = np.argsort(eigenvalues)
            eigenvalues = eigenvalues[order]
            full_vectors = full_vectors[:, order]

            omega = np.sqrt(eigenvalues)
            frequencies = omega / (2.0 * np.pi)
            valid = frequencies > self.EIGEN_TOL

            eigenvalues = eigenvalues[valid]
            full_vectors = full_vectors[:, valid]
            frequencies = frequencies[valid]
            periods = np.where(frequencies > self.EIGEN_TOL, 1.0 / frequencies, np.inf)

            count = min(n_modes, len(frequencies))
            full_vectors = full_vectors[:, :count]
            frequencies = frequencies[:count]
            periods = periods[:count]

            participation, eff_mass, modal_mass = self._participation(
                full_vectors, M_red, self.free_dofs,
            )
            total_mass_x, total_mass_y, total_mass_z = self._total_masses(
                M_red, self.free_dofs,
            )

        return ModalResult(
            frequencies=frequencies,
            periods=periods,
            mode_shapes=full_vectors,
            modal_dofs=self.free_dofs,
            modal_mass=modal_mass,
            participation_factor_x=participation[:, 0],
            participation_factor_y=participation[:, 1],
            participation_factor_z=participation[:, 2],
            effective_mass_x=eff_mass[:, 0],
            effective_mass_y=eff_mass[:, 1],
            effective_mass_z=eff_mass[:, 2],
            total_mass_x=total_mass_x,
            total_mass_y=total_mass_y,
            total_mass_z=total_mass_z,
        )

    def _build_effective_mass(self, M: np.ndarray) -> np.ndarray:
        m_diag = np.diag(M).astype(float).copy()
        n = len(m_diag)
        if n == 0:
            return M.copy()

        M_eff = M.copy()
        m_max = float(m_diag.max()) if n else 1.0
        mass_tol = 1e-9 * max(1.0, m_max)

        trans_mass_map: dict[int, float] = {}
        for i, dof in enumerate(self.free_dofs):
            dof_int = int(dof)
            if dof_int % 6 == 0 and m_diag[i] > mass_tol:
                trans_mass_map[dof_int] = float(m_diag[i])

        for i in range(n):
            dof_int = int(self.free_dofs[i])
            dof_type = dof_int % 6
            if dof_type in (3, 4, 5):
                node_base = dof_int - dof_type
                m_trans = trans_mass_map.get(node_base, 1000.0)
                if M_eff[i, i] < self.MIN_MASS:
                    M_eff[i, i] = max(m_trans * self.ROT_MASS_SCALE, self.MIN_MASS)

        return M_eff

    def _participation_condensed(
        self,
        vectors_a: np.ndarray,
        M_cond: np.ndarray,
        trans_idx: np.ndarray,
        rot_idx: np.ndarray,
        G: Optional[np.ndarray],
        modal_dofs: np.ndarray,
    ) -> Tuple[np.ndarray, np.ndarray, np.ndarray]:
        n_modes = vectors_a.shape[1]
        if n_modes == 0:
            z = np.zeros((0, 3))
            return z, z, np.zeros(0)

        r_full = self._direction_vectors(modal_dofs)
        r_a = r_full[:, trans_idx]
        if G is not None:
            r_b = r_full[:, rot_idx]
            r_cond = r_a + r_b @ G
        else:
            r_cond = r_a

        M_phi = M_cond @ vectors_a
        modal_mass = np.einsum("ij,ij->j", vectors_a, M_phi)
        safe_mass = np.where(modal_mass > 0, modal_mass, 1.0)

        L = vectors_a.T @ M_cond @ r_cond.T
        participation = L / safe_mass[:, None]
        eff_mass = (L ** 2) / safe_mass[:, None]

        return participation, eff_mass, modal_mass

    def _total_masses_condensed(
        self,
        M_cond: np.ndarray,
        trans_idx: np.ndarray,
        rot_idx: np.ndarray,
        G: Optional[np.ndarray],
        modal_dofs: np.ndarray,
    ) -> Tuple[float, float, float]:
        r_full = self._direction_vectors(modal_dofs)
        r_a = r_full[:, trans_idx]
        if G is not None:
            r_b = r_full[:, rot_idx]
            r_cond = r_a + r_b @ G
        else:
            r_cond = r_a

        totals = []
        for i in range(3):
            r_i = r_cond[i]
            totals.append(float(r_i @ M_cond @ r_i))
        return tuple(totals)  # type: ignore

    def _participation(
        self,
        mode_shapes: np.ndarray,
        M_red: np.ndarray,
        modal_dofs: np.ndarray,
    ) -> Tuple[np.ndarray, np.ndarray, np.ndarray]:
        n_modes = mode_shapes.shape[1]
        if n_modes == 0:
            z = np.zeros((0, 3))
            return z, z, np.zeros(0)

        r = self._direction_vectors(modal_dofs)
        M_phi = M_red @ mode_shapes
        modal_mass = np.einsum("ij,ij->j", mode_shapes, M_phi)
        safe_mass = np.where(modal_mass > 0, modal_mass, 1.0)

        L = mode_shapes.T @ M_red @ r.T
        participation = L / safe_mass[:, None]
        eff_mass = (L ** 2) / safe_mass[:, None]

        return participation, eff_mass, modal_mass

    @staticmethod
    def _total_masses(
        M_red: np.ndarray,
        modal_dofs: np.ndarray,
    ) -> Tuple[float, float, float]:
        d = np.asarray(modal_dofs) % 6
        rx = (d == 0).astype(float)
        ry = (d == 1).astype(float)
        rz = (d == 2).astype(float)
        return (
            float(rx @ M_red @ rx),
            float(ry @ M_red @ ry),
            float(rz @ M_red @ rz),
        )

    def _direction_vectors(self, modal_dofs: np.ndarray) -> np.ndarray:
        d = np.asarray(modal_dofs) % 6
        return np.vstack([
            (d == 0).astype(float),
            (d == 1).astype(float),
            (d == 2).astype(float),
        ])
```

```python
def run_isolated_single_beam_test(proj, P_load_N=1000.0):
    # Rijitlik matrisini kur ve çöz
    stiff_asm = dynamic_build_stiffness(proj, use_shear_deform=True)
    F = np.zeros(stiff_asm.num_dofs)
    F[node_mid.dof_indices[2]] = -P_load_N
    fixed_dofs = local_collect_fixed_dofs(proj.scene)
    solver = StructuralSolver(stiff_asm.K_global, np.zeros_like(stiff_asm.K_global), fixed_dofs, stiff_asm.num_dofs)
    U_ = solver.static_solve(F)
    u_numerical = abs(U_[node_mid.dof_indices[2]])
    return u_numerical

#TEST1
# 1. Proje, Sahne, Malzeme ve Kesit Tanımlamaları
proj = Project()
scn = proj.scene

# S275 Çelik Malzemesi
steel = Material(name='S275')
steel.E1 = 210000.0
steel.G12 = 80769.2307692308
steel.density = 7.84904737995992e-06

box200 = Section(name="BOX200x200x10", material=steel)
box200.profile_params = {
"Area": 200**2 - 180**2,
"J": 3.61e8,  # Gerçek torsiyon sabiti ayrıca hesaplanmalı.
"I33": (200**4 - 180**4) / 12,
"I22": (200**4 - 180**4) / 12,
"AS2": 0.0,  # Kesme testi yapılacaksa gerçek değer girilmeli.
"AS3": 0.0,
}

L = 6000.0  # Toplam Kiriş Boyu (mm)
E = steel.E1
I33 = box200.profile_params['I33']

# 2. Düğümlerin (Nodes) Oluşturulması
node_left = Node(x=0.0, y=0.0, z=0.0)
node_left.label = "LEFT_SUPPORT"
node_left.unique_id = 101
node_mid = Node(x=3000.0, y=0.0, z=0.0)
node_mid.label = "MID_TEST"
node_mid.unique_id = 102
node_right = Node(x=6000.0, y=0.0, z=0.0)
node_right.label = "RIGHT_SUPPORT"
node_right.unique_id = 103

# Kiriş Elemanlarının Tanımlanması
frame1 = Frame(node_i=node_left, node_j=node_mid, section=box200)
frame1.label = "BEAM_PART1"


frame2 = Frame(node_i=node_mid, node_j=node_right, section=box200)
frame2.label = "BEAM_PART2"


scn.add_node(node_left)
scn.add_node(node_right)
scn.add_node(node_mid)
scn.add_frame(frame1)
scn.add_frame(frame2)

plane_restraint = Restraint(ux=False, uy=True, uz=False, rx=True, ry=False, rz=True,)
node_mid.restraint = plane_restraint

# 3. SENARYO A: İKİ UCU TAM ANKASTRE MESNET KOŞULU
P_load_N=2500.0

restraint_fixed = Restraint(ux=True, uy=True, uz=True, rx=True, ry=True, rz=True)
node_left.restraint = restraint_fixed
node_right.restraint = restraint_fixed
result_num = run_isolated_single_beam_test(proj, P_load_N)
print(f"  ankastre - Sayısal Çözüm (Shear Dahil): {result_num:.6f} mm\n")
u_fixed_theoretical = (P_load_N * (L**3)) / (192 * E * I33)
print(f"  ankastre - Analitik Çözüm: {u_fixed_theoretical:.6f} mm\n")

restraint_pinned = Restraint(ux=True, uy=True, uz=True, rx=False, ry=False, rz=False)
node_left.restraint = restraint_pinned
node_right.restraint = restraint_pinned
result_num = run_isolated_single_beam_test(proj, P_load_N)
print(f"  pinned - Sayısal Çözüm (Shear Dahil): {result_num:.6f} mm\n")
u_simple_theoretical = (P_load_N * (L**3)) / (48 * E * I33)
print(f"  pinned - Analitik Çözüm: {u_simple_theoretical:.6f} mm\n")


# TEST2

# 1. Proje, Sahne, Malzeme ve Kesit Tanımlamaları
proj = Project()
scn = proj.scene

# S275 Çelik Malzemesi
steel = Material(name='S275')
steel.E1 = 210000.0
steel.G12 = 80769.2307692308
steel.density = 7.84904737995992e-06

# IPE200 Kesiti
ipe200 = Section(name='IPE200', material=steel)
ipe200.profile_params = {
    'Area': 2850.0,
    'J': 69200.0,
    'I33': 19430000.0,
    'I22': 1420000.0,
    'AS2': 1120.0,
    'AS3': 1416.667
}

L = 6000.0  # Toplam Kiriş Boyu (mm)
E = steel.E1
I33 = box200.profile_params['I33']

# 2. Düğümlerin (Nodes) Oluşturulması
node_left = Node(x=0.0, y=0.0, z=0.0)
node_left.label = "LEFT_SUPPORT"
node_left.unique_id = 101
node_mid = Node(x=3000.0, y=0.0, z=0.0)
node_mid.label = "MID_TEST"
node_mid.unique_id = 102
node_right = Node(x=6000.0, y=0.0, z=0.0)
node_right.label = "RIGHT_SUPPORT"
node_right.unique_id = 103

# Kiriş Elemanlarının Tanımlanması
frame1 = Frame(node_i=node_left, node_j=node_mid, section=ipe200)
frame1.label = "BEAM_PART1"


frame2 = Frame(node_i=node_mid, node_j=node_right, section=ipe200)
frame2.label = "BEAM_PART2"


scn.add_node(node_left)
scn.add_node(node_right)
scn.add_node(node_mid)
scn.add_frame(frame1)
scn.add_frame(frame2)

plane_restraint = Restraint(ux=False, uy=True, uz=False, rx=True, ry=False, rz=True,)
node_mid.restraint = plane_restraint

# 3. SENARYO A: İKİ UCU TAM ANKASTRE MESNET KOŞULU
P_load_N=2500.0

restraint_fixed = Restraint(ux=True, uy=True, uz=True, rx=True, ry=True, rz=True)
node_left.restraint = restraint_fixed
node_right.restraint = restraint_fixed
result_num = run_isolated_single_beam_test(proj, P_load_N)
print(f"  ankastre - Sayısal Çözüm (Shear Dahil): {result_num:.6f} mm\n")
u_fixed_theoretical = (P_load_N * (L**3)) / (192 * E * I33)
print(f"  ankastre - Analitik Çözüm: {u_fixed_theoretical:.6f} mm\n")

restraint_pinned = Restraint(ux=True, uy=True, uz=True, rx=False, ry=False, rz=False)
node_left.restraint = restraint_pinned
node_right.restraint = restraint_pinned
result_num = run_isolated_single_beam_test(proj, P_load_N)
print(f"  pinned - Sayısal Çözüm (Shear Dahil): {result_num:.6f} mm\n")
u_simple_theoretical = (P_load_N * (L**3)) / (48 * E * I33)
print(f"  pinned - Analitik Çözüm: {u_simple_theoretical:.6f} mm\n")

```
